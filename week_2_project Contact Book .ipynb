```python
# Main dictionary to store contact details
phonebook = {}

while True:
    print("\n-------------------------")
    print("   MY CONTACT BOOK")
    print("-------------------------")
    print("1. Add New Contact")
    print("2. Find Contact")
    print("3. Exit")

    user_choice = input("Select an option (1-3): ").strip()

    # 1. Add Contact Logic
    if user_choice == "1":
        person_name = input("Enter Contact Name: ").strip().title()
        phone_num = input("Enter Phone Number: ").strip()
        email_addr = input("Enter Email Address: ").strip()

        # Check if any input field is empty
        if not person_name or not phone_num or not email_addr:
            print(">> Error: All fields (Name, Phone, Email) are mandatory!")
        else:
            # Store details in nested format
            phonebook[person_name] = {"phone": phone_num, "email": email_addr}
            print(f">> Success: Details for '{person_name}' saved successfully!")

    # 2. Search Contact Logic
    elif user_choice == "2":
        search_query = input("Enter Name to Search: ").strip().title()

        if search_query in phonebook:
            info = phonebook[search_query]
            print(f"\n[ Contact Found ]")
            print(f"Name  : {search_query}")
            print(f"Phone : {info['phone']}")
            print(f"Email : {info['email']}")
        else:
            print(f">> Sorry, no contact found for '{search_query}'.")

    # 3. Exit Logic
    elif user_choice == "3":
        print(">> Closing Contact Book. Have a great day!")
        break

    # Invalid Option Handling
    else:
        print(">> Invalid choice! Please enter only 1, 2, or 3.")
```
